```python
#sabemos que hay años que tiene valores nulos. Vamos a comprobar y limpiar mas abajo:
df[df["Year"].isna()].head(10)
```

```python
#sabemos que publisher tiene valores nulos vamos a comprobar:
df[df["Publisher"].isna()].head(10)
```

```python
#Lo que vamos a hacer es crear una copia y trabajar sobre la copia para borrar los nulos
df_limpio = df.copy()
#cogemos columna publisher y en vez de que salga nulo va a salir unknown
df_limpio["Publisher"] = df_limpio["Publisher"].fillna("Unknown")
#dropna elimnina las filas con hueco y subset limita la búsqueda a la columna Year
df_limpio = df_limpio.dropna(subset=["Year"])
#.astype(int) convierte los años de decimal (2006.0) a entero (2006).
df_limpio["Year"] = df_limpio["Year"].astype(int)
print("Filas antes :", len(df))
print("Filas ahora :", len(df_limpio))
df_limpio.isna().sum()
```
